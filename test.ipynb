```python
import typing as t
import pydantic as pdt

class BandsInput(pdt.BaseModel):
    model_config = pdt.ConfigDict(
        title="Band Structure",
    )
    energy_cutoff: t.Annotated[
        float,
        pdt.Field(
            40.0,
            ge=0,
            le=500,
            multiple_of=0.1,
            description="Energy cutoff (Ry)",
        ),
    ]
    kpoint_distance: t.Annotated[
        float,
        pdt.Field(
            0.1,
            gt=0,
            le=1,
            multiple_of=0.1,
            description="K-point distance (1/Angstrom)",
        ),
    ]
```
